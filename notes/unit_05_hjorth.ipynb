Hjorth 特征的全称一般是：

**Hjorth Parameters**

中文常译为：

**Hjorth 参数** 或 **Hjorth 时域特征**

它主要包括 3 个经典特征：

| 特征         | 英文全称              | 中文含义 |
| ---------- | ----------------- | ---- |
| Activity   | Hjorth Activity   | 活动度  |
| Mobility   | Hjorth Mobility   | 移动度  |
| Complexity | Hjorth Complexity | 复杂度  |

它们是 Bo Hjorth 提出的 EEG 时域特征，用来描述信号的幅度变化、频率变化和波形复杂程度。

简单理解：

```text
Activity   → 信号整体能量 / 方差
Mobility   → 信号变化快慢 / 主频趋势
Complexity → 信号形状复杂程度
```

常见写法：

```python
hjorth_activity
hjorth_mobility
hjorth_complexity
```

或者：

```python
activity, mobility, complexity = hjorth_parameters(signal)
```
